```python
import pandas as pd
from collections import Counter
import matplotlib.pyplot as plt
import seaborn as sns
```

## Persons

```python
persons_edges = pd.read_csv('persons_edges.csv')
persons_nodes = pd.read_csv('persons_nodes.csv')
```

```python
authors = dict()
persons = dict()

for i,row in persons_nodes.iterrows():
    if str(row['Id']).strip().startswith('\'AUTH'):
        authors[ row['Id'] ] = row['Label']
    else:
        persons[ row['Id'] ] = row['Label']
```

```python
for author in authors:
    print(authors[author])
    persons_author = persons_edges[ persons_edges.Source == author ]
    persons_count = Counter()
    for i,row in persons_author.iterrows():
        persons_count.update([ persons[ row['Target'] ] ] ) 
    
    x_axis = []
    y_axis = []
    for name,count in persons_count.most_common():
        x_axis.append(name)
        y_axis.append(count)
        
    
    fig = plt.figure( figsize=( 10 , 5 ) )

    colours = [ '#c7c4af' , '#0e9e30', '#a0061a' , '#6a55d4' , '#cf8817'  ] 



    graph = sns.barplot( x=x_axis, y= y_axis, palette = colours )

    graph.set_title( authors[author] , size = 20) 
    graph.set_xlabel( 'Correspondent' , size = 14) 
    graph.set_ylabel('$ references' , size = 14 )

    plt.xticks(rotation= 80)
    # The next line places the legend outside out the plot
    plt.legend( bbox_to_anchor=(1.05, 1),loc=2, borderaxespad=0.);


    plt.show()
        
    
        
```
