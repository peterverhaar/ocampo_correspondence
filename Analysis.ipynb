```python
import pandas as pd
import os

import nltk
from nltk import word_tokenize,pos_tag,sent_tokenize

from collections import Counter
#from langdetect import detect

from text_mining import *
```

```python
dir = 'TXT'
correspondence = os.listdir(dir)
correspondence = [os.path.join(dir,file) for file in correspondence if re.search(r'txt$',file)]
```

## Unique words

```python
for author in correspondence:
    print(f'\n{author}\n')
    text1 = ''
    text2 = ''

    file = open(author,encoding='utf-8')
    text1 = file.read().lower()
    words1 = word_tokenize(text1)
    file.close()

    for other_author in correspondence:
        if other_author != author:
            file = open(other_author,encoding='utf-8')
            text2 += file.read().lower()
            file.close()

    words2 = word_tokenize(text2)

    unique_words = []
    for word in words1:
        if word not in words2:
            unique_words.append(word)

    unique = Counter(unique_words)

    for word,count in unique.most_common(40):
        print(word,count)
```

## Distinctive words

```python
for author in correspondence:

    print(f'\n{author}\n')

    corpus1 = []
    corpus2 = []

    for text in correspondence:
        if text == author:
            corpus1.append(text)
        else:
            corpus2.append(text)


    freq1 = calculate_word_frequencies(corpus1)
    total1 = sum(freq1.values())

    freq2 = calculate_word_frequencies(corpus2)
    total2 = sum(freq2.values())

    ll_scores = dict()


    for word in freq1.keys():
        if word in freq2.keys():

            ll_score = log_likelihood( freq1[word] , freq2[word] , total1 , total2 )
            ll_scores[word] = ll_score

    max = 40
    i = 0

    for word in sorted_by_value(ll_scores , ascending = False ):
        print( word , ll_scores[word] )
        i += 1
        if i == max:
            break
```

```python
def find_most_frequent_NE(code,document):
  ne = []
  for named_entity in document.ents:
    if named_entity.label_ == code and re.search(r'[A-Za-z]{2,}',str(named_entity)):
      ne.append(str(named_entity).strip())
  return Counter(ne)


def get_ner_in_context(keyword, document, desired_ner_labels= False):

    if desired_ner_labels != False:
        desired_ner_labels = desired_ner_labels
    else:
         # all possible labels
        desired_ner_labels = list(nlp.get_pipe('ner').labels)

    #Iterate through all the sentences in the document and pull out the text of each sentence
    for sentence in document.sents:
        #process each sentence
        sentence_doc = nlp(sentence.text)
        for named_entity in sentence_doc.ents:
            #Check to see if the keyword is in the sentence (and ignore capitalization by making both lowercase)
            if keyword.lower() in named_entity.text.lower()  and named_entity.label_ in desired_ner_labels:
                #Use the regex library to replace linebreaks and to make the keyword bolded, again ignoring capitalization
                #sentence_text = sentence.text

                sentence_text = re.sub('\n', ' ', sentence.text)
                sentence_text = re.sub(f"{named_entity.text}", f"**{named_entity.text}**", sentence_text, flags=re.IGNORECASE)

                print('---')
                display(Markdown(f"**{named_entity.label_}**"))
                display(Markdown(sentence_text))



def find_sentences_with_keyword(keyword, document):

    #Iterate through all the sentences in the document and pull out the text of each sentence
    for sentence in document.sents:
        sentence = sentence.text

        #Check to see if the keyword is in the sentence (and ignore capitalization by making both lowercase)
        if keyword.lower() in sentence.lower():

            #Use the regex library to replace linebreaks and to make the keyword bolded, again ignoring capitalization
            sentence = re.sub('\n', ' ', sentence)
            sentence = re.sub(f"{keyword}", f"**{keyword}**", sentence, flags=re.IGNORECASE)

            display(Markdown(sentence))

nlp = spacy.load('es_core_news_md')
```

```python
for text in correspondence:
  print('\n')
  print(display(Markdown(f'**{text.upper()}**')))

  full_text = open(correspondence[0], encoding='utf-8').read()
  document = nlp(full_text)

  print('\n---- Most frequent adjectives ----')

  adjectives = []

  for token in document:
    if token.pos_ == 'ADJ' and re.search(r'[A-Za-z]{2,}',token.lemma_):
      adjectives.append(token.lemma_)

  adjectives_freq = Counter(adjectives)

  for word,count in adjectives_freq.most_common(40):
      print(f'{word} {count}' )

  print('\n---- Most frequent nouns ----')

  nouns = []

  for token in document:
    if token.pos_ == 'NOUN' and re.search(r'[A-Za-z]{2,}',token.lemma_):
      adjectives.append(token.lemma_)

  nouns_freq = Counter(adjectives)

  for word,count in nouns_freq.most_common(40):
      print(f'{word} {count}' )

  print('\n---- Sentences containing a specific word ----')

  find_sentences_with_keyword(keyword="escrito", document=document)

  print('\n---- Most frequent personal names ----')

  ne_freq = find_most_frequent_NE('PER',document)
  for ne,count in ne_freq.most_common(30):
    print(f'{ne} => {count}')

  print('\n---- Most frequent locations ----')

  ne_freq = find_most_frequent_NE('LOC',document)
  for ne,count in ne_freq.most_common(30):
    print(f'{ne} => {count}')

  get_ner_in_context('Victoria Ocampo', document)

  ana = SentimentIntensityAnalyzer()

  positive_words = []
  negative_words = []
  total = 0

  for sentence in document.sents:
      words = word_tokenize(str(sentence).lower())
      words = [word for word in words if len(word)>2]
      total += len(words)
      for word in words:
          scores = ana.polarity_scores(word)
          if scores["pos"] > 0.8:
              positive_words.append(word)
          elif scores["neg"] > 0.8:
              negative_words.append(word)

  print('\n---- Most frequent positive words ----')

  sentiment_freq = Counter(positive_words)

  for word,count in sentiment_freq.most_common(20):
      print(word)

  print('\nPercentage of positive words:')

  print(sum(sentiment_freq.values())/total)


  print('\nNegative words:')

  sentiment_freq = Counter(negative_words)

  for word,count in sentiment_freq.most_common(20):
      print(word)

  print('\n---- Most frequent negative words ----')

  print(sum(sentiment_freq.values())/total)

```
